```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("heritage_sites.csv")


df.head()
```

```python
# Count the number of heritage sites in each country
country_counts = df["Country"].value_counts()

# bar chart
plt.figure(figsize=(10, 6))
country_counts.plot(kind="bar", color="#FFC0CB")

plt.title("Number of Heritage Sites by Country")
plt.xlabel("Country")
plt.ylabel("Number of Heritage Sites")
plt.tight_layout()

plt.show()
```

Italy has the largest number of heritage sites in this dataset, followed by 
the United Kingdom, France, Belgium, and Austria. Some countries, such as 
Croatia and Czechia, have only one site represented. This shows that the 
dataset is not evenly distributed across countries, with some countries 
having a much larger representation of heritage sites.

```python
# Convert the columns to numbers
df["Year of Inscription"] = pd.to_numeric(df["Year of Inscription"], errors="coerce")
df["Area (ha)"] = pd.to_numeric(df["Area (ha)"], errors="coerce")

# scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(df["Year of Inscription"], df["Area (ha)"])

plt.title("Heritage Site Area vs. Year of Inscription")
plt.xlabel("Year of Inscription")
plt.ylabel("Area (hectares)")

plt.show()
```

The scatter plot does not show a clear relationship between the year a site 
was inscribed and its area. Large and small heritage sites were added to the 
World Heritage List throughout the different years. There are several 
particularly large sites, but they do not appear to be concentrated in one 
specific period.

```python
# Count the number of sites for each historical period
period_counts = df["Historical Period"].value_counts()

# bar chart
plt.figure(figsize=(12, 6))
period_counts.plot(kind="bar")

plt.title("Number of Heritage Sites by Historical Period")
plt.xlabel("Historical Period")
plt.ylabel("Number of Heritage Sites")

plt.tight_layout()
plt.show()
```

Medieval sites are the most common historical period in this dataset. Several 
sites also combine multiple periods, such as Medieval–Modern or 
Medieval–Renaissance. This suggests that many of the heritage sites represent 
long periods of historical development rather than belonging to only one 
specific period.
