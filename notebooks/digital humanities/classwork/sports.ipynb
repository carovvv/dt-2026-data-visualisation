```python
# can use to work with tabular data
import pandas as pd
df = pd.read_csv("competition.csv")
df.head() # show the first few rows
df.describe() #show basic statistics
```

```python
import matplotlib.pyplot as plt # matplotlib helps create charts and visualaitations
df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False) #legend takes away what the color shows
#bar means the type of chart you would want and you put x and y as the data you want to be shown from your data
plt.title("Years of Practice per athlete") #adds a title
plt.ylabel("Years of Practice") #creates a title for y axis
plt.show()
```

```python
import matplotlib.pyplot as plt 
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.ylabel("Age")
```

```python
df.plot(kind="pie", y="Age", labels=df["Name"], legend=False, autopct="%1.1f%%", colormap="viridis")
plt.title("Age of athletes")
```

```python
plt.scatter(df["Age"], df["Years_of_practice"],s=df["Years_of_practice"]*60, alpha=0.6)
#s part means if student has more ýears of practice the bubble is bigger
plt.title("Years of practice by age")
plt.ylabel("Years of practice")
plt.xlabel("Age")
```
