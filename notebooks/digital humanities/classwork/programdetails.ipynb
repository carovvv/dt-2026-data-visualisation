```python
import pandas as pd
df= pd.read_csv("comp.csv")
df
```

```python
import matplotlib.pyplot as plt

hours_by_student = df.groupby("Student")["Hours_Studied"].mean()

hours_by_student.plot(kind="bar", color="#FFC0CB")

plt.xlabel("Student")
plt.ylabel("Total Hours Studied")
plt.title("Total Study Hours by Student")


plt.show()
```

```python
plt.scatter(df["Hours_Studied"],df["Grade"],s=df["Grade"] * 0.5, alpha=0.6,  color="#3BE77A")

plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.title("Hours Studied vs Grade")
plt.show()


```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

df[["Hours_Studied", "Grade"]].boxplot()

plt.title("Hours Studied and Grades")
plt.ylabel("Value")

plt.show()
```
